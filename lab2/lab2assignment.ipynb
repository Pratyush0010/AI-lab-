# Linear Regression

### 1) Data Retrieval and Collection

```python
# Import pandas
import pandas as pd
```

```python
# Load the dataset
df = pd.read_csv("Salary_Data.csv")
```

```python
# Display shape of dataset
print("Shape of dataset:", df.shape)
# Display column names
print("Column names:", df.columns.tolist())
```

```python
# Basic information about the dataset
print(df.info())
```

### 2) Data Cleaning

```python
# Check for missing values
print("Missing values in each column:")
print(df.isnull().sum())
```

```python
# Verify data types
print("Data types:")
print(df.dtypes)
```

### 3)Feature Design

```python
# Separate feature and target variable
X = df[['YearsExperience']]   # Feature
y = df['Salary']              # Label
```

```python
print("Feature (X):")
print(X.head())
print("\nLabel (y):")
print(y.head())
```

### 4) Algorithm Selection

###### I chose Linear Regression for this task because the thing we are trying to predict, which is Salary is a number that can be anything, not a few possibilities. The information we have to work with is also pretty simple we just have one thing to look at which's YearsExperience. Linear Regression is good at showing how one or more things can affect something by drawing a straight line through the data. We know that usually the more years of experience someone has the higher their Salary will be. So it makes sense to use Linear Regression because we think there is a line relationship between Salary and YearsExperience. Linear Regression is a choice, for this because it can help us understand how Salary changes when YearsExperience changes.

### 5) Loss Function Selection

##### Mean Squared Error (MSE):

##### MSE = (1/n) · Σ(yᵢ - ŷᵢ)²

###### The Mean Squared Error was chosen to see how good the model is. Mean Squared Error looks at the average of the differences, between what happened and what the model said would happen. It does this by squaring the differences big mistakes are punished more. This helps the model get better at making predictions. When the Mean Squared Error is low that means the model is doing a job.

### 6) Model Learning and Training

```python
# Import Linear Regression
from sklearn.linear_model import LinearRegression

# Create the model
model = LinearRegression()

# Train the model
model.fit(X, y)
print("Model training completed successfully.")
```

```python
# Model parameters
print("Coefficient (Slope):", model.coef_[0])
print("Intercept:", model.intercept_)
```

### 7) Model Evaluation

```python
from sklearn.metrics import mean_squared_error, r2_score

# Predictions
y_pred = model.predict(X)

# Evaluation metrics
mse = mean_squared_error(y, y_pred)
r2 = r2_score(y, y_pred)

print("Mean Squared Error (MSE):", mse)
print("R² Score:", r2)
```

##### Interpretation:
###### The obtained MSE value indicates the average prediction error of the model. The R² score shows how well the variation in salary is explained by years of experience. A high R² score and a low MSE suggest that the Linear Regression model fits the dataset well and can make reliable salary predictions.

### Visualization

```python
import matplotlib.pyplot as plt

# Predictions
y_pred = model.predict(X)
```

```python
# Scatter plot of actual data
plt.scatter(X, y, label='Actual Data')

# Regression line
plt.plot(X, y_pred, label='Regression Line')

plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.title('Salary vs Years of Experience')
plt.legend()
plt.show()
```

### Predicted vs Actual Values

```python
comparison = pd.DataFrame({
    'Actual Salary': y,
    'Predicted Salary': y_pred
})
comparison.head(10)
```

### Assumptions of Linear Agression

###### 1) Linearity – There is a linear relationship between the independent variable (YearsExperience) and the dependent variable (Salary).
###### 2) Independence of Observations – The observations in the dataset are independent of one another.
###### 3) Homoscedasticity – The variance of the residuals (prediction errors) remains constant across all levels of the independent variable.
###### 4) Normality of Residuals – The residuals are approximately normally distributed.
###### 5) No Significant Outliers – Extreme outliers can disproportionately influence the regression line and reduce model accuracy.
