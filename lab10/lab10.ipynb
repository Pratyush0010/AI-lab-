## Theory

BLEU (Bilingual Evaluation Understudy) is an automatic evaluation metric used in Artificial Intelligence and Natural Language Processing (NLP) to measure the quality of machine-translated text. It compares a candidate sentence generated by a model with one or more reference sentences written by humans. A higher BLEU score indicates that the generated translation is more similar to the reference translation.

BLEU evaluates translations using *n-grams*, which are sequences of consecutive words. It calculates the precision of matching n-grams while also applying a brevity penalty to discourage excessively short translations.

### BLEU-1

BLEU-1 uses **1-grams (individual words)** to evaluate translations. It measures how many words in the candidate sentence are present in the reference sentence. This metric focuses on word-level accuracy without considering word order.

The BLEU-1 score is calculated as:

> BLEU-1 = (Number of matching 1-grams) / (Total number of 1-grams in the candidate sentence)

A higher BLEU-1 score indicates that the generated translation contains words that closely match the reference translation.

### BLEU-2

BLEU-2 uses **2-grams (pairs of consecutive words)** for evaluation. Unlike BLEU-1, it considers both word selection and the local ordering of words. This makes BLEU-2 more effective at determining whether the generated sentence preserves the phrase structure of the reference sentence.

The BLEU-2 score is calculated as:

> BLEU-2 = (Number of matching 2-grams) / (Total number of 2-grams in the candidate sentence)

Since BLEU-2 evaluates word pairs, it provides a better indication of fluency and phrase-level correctness than BLEU-1.

### Importance of BLEU Scores

* BLEU-1 evaluates word-level similarity between the generated and reference sentences.
* BLEU-2 evaluates phrase-level similarity by considering consecutive word pairs.
* Higher BLEU scores generally indicate better translation quality.
* BLEU metrics are widely used for evaluating machine translation systems and other text generation models in NLP.
* BLEU-2 is usually more informative than BLEU-1 because it considers both word correctness and partial word ordering.


#### Import Libraries

```python
import pandas as pd
from nltk.translate.bleu_score import sentence_bleu
from nltk.translate.bleu_score import SmoothingFunction
```

#### Read Dataset

```python
english_sentences = []
french_sentences = []

with open("eng-fra.txt", "r", encoding="utf-8") as file:
    for line in file:
        parts = line.strip().split("\t")

        # Some datasets have more than 2 columns.
        if len(parts) >= 2:
            english_sentences.append(parts[0])
            french_sentences.append(parts[1])

print("Total sentences :", len(english_sentences))
```

#### Create Candidate Sentences

```python
candidate_sentences = french_sentences.copy()
```

#### Calculate BLEU 1 Scores

```python
smooth = SmoothingFunction().method1

bleu1_scores = []

for reference, candidate in zip(french_sentences,
                                candidate_sentences):

    reference_tokens = [reference.split()]
    candidate_tokens = candidate.split()

    score = sentence_bleu(
        reference_tokens,
        candidate_tokens,
        weights=(1,0,0,0),
        smoothing_function=smooth
    )

    bleu1_scores.append(score)

average_bleu1 = sum(bleu1_scores)/len(bleu1_scores)

print("Average BLEU-1 Score :", average_bleu1)
```

#### Calculate BLEU 2 Scores

```python
bleu2_scores = []

for reference, candidate in zip(french_sentences,
                                candidate_sentences):

    reference_tokens = [reference.split()]
    candidate_tokens = candidate.split()

    score = sentence_bleu(
        reference_tokens,
        candidate_tokens,
        weights=(0.5,0.5,0,0),
        smoothing_function=smooth
    )

    bleu2_scores.append(score)

average_bleu2 = sum(bleu2_scores)/len(bleu2_scores)

print("Average BLEU-2 Score :", average_bleu2)
```

## Discussion

BLEU-1 and BLEU-2 were used to evaluate the similarity between candidate and reference translations in the English–French dataset. BLEU-1 measures word-level accuracy, while BLEU-2 considers both word accuracy and word order using two-word sequences. Higher BLEU scores indicate better translation quality.

## Conclusion

This experiment demonstrated the use of BLEU-1 and BLEU-2 for machine translation evaluation. BLEU-1 provides word-level evaluation, whereas BLEU-2 provides phrase-level evaluation, making both metrics useful for assessing translation performance.
